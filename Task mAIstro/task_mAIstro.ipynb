## Teste do Grafo

```python
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import MemorySaver
from src.graph import builder

# Compila com MemorySaver local para testes com thread_id no Notebook
local_graph = builder.compile(checkpointer=MemorySaver())

# Configuração da thread
config = {"configurable": {"thread_id": "conversa_jack_1"}}

# Mensagem de entrada inicial
input_state = {
    "messages": [
        #HumanMessage(content="Olá! Meu nome é Jackson e sou desenvolvedor python."),
        #HumanMessage(content="Quero que você sempre salve a data da adição da tarefa quando for adicionar uma nova tarefa."),
        #HumanMessage(content="Por favor, adicione as tarefas 'Comprar leite' e 'Comprar Pão' à minha lista de tarefas."),
        HumanMessage(content="Quem sou eu? Quais são minhas preferências e quais são minhas tarefas?"),
    ]
}

# Execução do grafo
result = local_graph.invoke(input_state, config=config)

print("--- Resposta do Agente ---")
print(result["messages"][-1].content)

print("\n--- Memória de Perfil ---")
print(result.get("profile"))

print("\n--- Memória de Tarefas ---")
print(result.get("tasks"))

print("\n--- Memória de Preferências ---")
print(result.get("preferences"))
```
